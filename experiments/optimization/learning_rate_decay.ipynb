```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
import matplotlib as plt
```

```python
x,y=make_classification(n_samples=2000,n_features=30,n_informative=20,n_redundant=5,n_classes=5,random_state=123)

x_train,x_temp,y_train,y_temp=train_test_split(x,y,test_size=0.3,random_state=27,stratify=y)
x_val,x_test,y_val,y_test=train_test_split(x_temp,y_temp,test_size=0.5,random_state=33,stratify=y_temp)
```

```python
rng=np.random.default_rng(123)
input_size=30
hidden_size=32
output_size=5

w1=rng.normal(0,0.1,size=(input_size,hidden_size))
b1=np.zeros(hidden_size)

w2=rng.normal(0,0.1,size=(hidden_size,output_size))
b2=np.zeros(output_size)

ilr=0.01
dr=0.005
```

```python
def relu(z):
    return np.maximum(0,z)

def softmax(z):
    zs=z-np.max(z,axis=1,keepdims=True)
    exp_zs=np.exp(zs)
    return exp_zs/np.sum(exp_zs,axis=1,keepdims=True)
def loss_calc(p,y):
    true_p=p[np.arange(len(y)),y]
    return -np.mean(np.log(true_p))
def z_calc(x,w,b):
    return x @ w + b
```

```python
for epoch in range(1000):
    z=z_calc(x_train,w1,b1)
    z_relu=relu(z)
    oz=z_calc(z_relu,w2,b2)
    p=softmax(oz)

    loss=loss_calc(p,y_train)

    err=p.copy()
    err[np.arange(len(y_train)),y_train]-=1
    err/=len(y_train)

    dw2=z_relu.T @ err
    db2=np.sum(err,axis=0)

    dh=err @ w2.T
    dh[z<=0]=0

    dw1=x_train.T @ dh
    db1=np.sum(dh,axis=0)

    lr = ilr/(1+dr*epoch)
    
    w1-=lr * dw1
    w2-=lr * dw2
    b1-=lr * db1
    b2-=lr * db2

    if epoch % 100==0:
        print(epoch,loss)
```
